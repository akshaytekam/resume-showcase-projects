### KPI — Total Revenue

```python
%sql
SELECT
    ROUND(SUM(total_revenue), 2) AS total_revenue
FROM retail_lakehouse.gold.gold_revenue;
```

### KPI — Total Orders

```python
%sql
SELECT
    SUM(total_orders) AS total_orders
FROM retail_lakehouse.gold.gold_revenue;
```

### KPI — Total Units

```python
%sql
SELECT
    SUM(total_units) AS total_units
FROM retail_lakehouse.gold.gold_revenue;
```

### KPI — Average Order Value

```python
%sql
SELECT
    ROUND(
        SUM(total_revenue) / SUM(total_orders),
        2
    ) AS average_order_value
FROM retail_lakehouse.gold.gold_revenue;
```

### Revenue trend

```python
%sql
SELECT
    order_date,
    ROUND(total_revenue, 2) AS revenue
FROM retail_lakehouse.gold.gold_revenue
ORDER BY order_date;
```

### Orders trend

```python
%sql
SELECT
    order_date,
    total_orders
FROM retail_lakehouse.gold.gold_revenue
ORDER BY order_date;
```

### Top 10 customers

```python
%sql
SELECT
    customer_id,
    total_orders,
    ROUND(total_spend, 2) AS total_spend,
    last_order_date
FROM retail_lakehouse.gold.gold_customers
ORDER BY total_spend DESC
LIMIT 10;
```

### Customer revenue distribution

```python
%sql
SELECT
    CASE
        WHEN total_spend >= 20000 THEN 'VIP'
        WHEN total_spend >= 10000 THEN 'HIGH_VALUE'
        WHEN total_spend >= 5000 THEN 'MEDIUM_VALUE'
        ELSE 'REGULAR'
    END AS customer_segment,

    COUNT(*) AS customers,
    ROUND(SUM(total_spend), 2) AS revenue

FROM retail_lakehouse.gold.gold_customers

GROUP BY
    CASE
        WHEN total_spend >= 20000 THEN 'VIP'
        WHEN total_spend >= 10000 THEN 'HIGH_VALUE'
        WHEN total_spend >= 5000 THEN 'MEDIUM_VALUE'
        ELSE 'REGULAR'
    END

ORDER BY revenue DESC;
```
